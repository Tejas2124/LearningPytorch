```python
X_train,x_test,Y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,:1],test_size=0.2,random_state=42)
```

```python
encoder = LabelEncoder()
Y_train = encoder.fit_transform(Y_train)
y_test = encoder.fit_transform(y_test)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
x_test = scaler.fit_transform(x_test)
```

```python
X_train_tensor = torch.from_numpy(X_train)
x_test_tensor = torch.from_numpy(x_test)
Y_train_tensor = torch.from_numpy(Y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Define Neural Network

```python
class simpleneural():
  def __init__(self,x):
    self.weights = torch.rand(x.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias = torch.rand(1,dtype=torch.float64,requires_grad=True)

  def forward(self,x):
    return torch.sigmoid(torch.matmul(x,self.weights) + self.bias)

  def loss_fxn(self,y_pred,y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(Y_train_tensor * torch.log(y_pred) + (1 - Y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

### Important Parameters

```python
learning_rate = 0.05
epochs = 20000
```

### training pipeline

```python

#create model
model = simpleneural(X_train_tensor)
model.weights
model.bias

#define loop
total_loss= 0
for epoch in range(epochs):
  #forward pass
  y_pred =model.forward(X_train_tensor)
  # print(f'y_pred : {y_pred}')
  # print(f'y_train : {Y_train_tensor}')
   #loss calculations
  loss = model.loss_fxn(y_pred,Y_train_tensor)
  # print(f'loss : {loss}')
  total_loss += loss
  #backward pass
  loss.backward()
  #weights/parameters updates
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad
  #zero gradient
  model.weights.grad.zero_()
  model.bias.grad.zero_()
  #print loss in each epochs
  print(f'loss at epoch {epoch+1} is {loss.item()}')
```

## Model evaluation

```python
with torch.no_grad():
  y_pred = model.forward(x_test_tensor)
  y_pred_class = (y_pred > 0.6).float()
  accuracy = (y_pred_class == y_test_tensor).float().mean()
  print(f'accuracy : {accuracy}')
```
