```python
import torch.nn as nn
import torch
```

```python
class model(nn.Module):

  def __init__(self,num_features):
    super().__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out

```

```python
#create dataset

features = torch.rand(10,5)


model = model(features.shape[1])

model(features)
```

## Using Sequential container

```python
class sequential_model(nn.Module):
  def __init__(self,num_features):
    super().__init__()

    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()
    # self.network = nn.Sequential(
    #     nn.Linear(num_features,4),
    #     nn.ReLU(),
    #     nn.Linear(4,2),
    #     nn.ReLU(),
    #     nn.Linear(2,1),
    #     nn.Sigmoid())

  def forward(self,features):
    return self.sigmoid(self.linear(features))
```

```python
model = sequential_model(features.shape[1])
```

## Dataset loading

```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.drop(columns = ['id','Unnamed: 32'],inplace=True)
```

```python
X_train,x_test,Y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,:1],test_size=0.2,random_state=42)
```

```python
encoder = LabelEncoder()
Y_train = encoder.fit_transform(Y_train)
y_test = encoder.fit_transform(y_test)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
x_test = scaler.fit_transform(x_test)
```

```python
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
x_test_tensor = torch.tensor(x_test, dtype=torch.float32)
Y_train_tensor = torch.tensor(Y_train, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32)
```

```python
print(X_train_tensor.dtype)
print(x_test_tensor.dtype)
print(Y_train_tensor.dtype)
print(y_test_tensor.dtype)
```

## important parameters

```python
learning_rate = 0.025
epochs = 50000
```

## USING TRAINING PIPELINE

```python
loss_function = nn.BCELoss()
```

```python
model = sequential_model(X_train_tensor.shape[1])
```

```python
optimizer = torch.optim.SGD(model.parameters(),lr=learning_rate)
```

```python
for epoch in range(epochs):

  #forward pass
  y_pred = model(X_train_tensor)

  #loss calculation
  loss=loss_function(y_pred,Y_train_tensor.view(-1,1))

  #zero gradients
  optimizer.zero_grad()


  #backward pass
  loss.backward()

  #update weights
  # with torch.no_grad():
  #   model.linear.weight -= learning_rate * model.linear.weight.grad
  #   model.linear.bias -= learning_rate * model.linear.bias.grad

  optimizer.step()

  # model.linear.weight.grad.zero_()
  # model.linear.bias.grad.zero_()
  print(f'loss at epoch {epoch+1} is {loss.item()}')
```

```python
with torch.no_grad():
  y_pred = model.forward(x_test_tensor)
  y_pred_class = (y_pred > 0.6).float()
  accuracy = (y_pred_class == y_test_tensor).float().mean()
  print(f'accuracy : {accuracy}')
```
