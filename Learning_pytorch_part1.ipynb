## Tensor Operation on GPU

```python
torch.cuda.is_available()
```

```python
device = torch.device('cuda')
```

```python
device
```

```python
#create new tensor on GPU
torch.rand((5,3),device = device)
```

```python
y = torch.rand(2,3)
y
```

```python
w= y.to(device=device)
```

```python
w + 5
```

```python
size = 20000
import time
```

```python
mat1 = torch.rand(size,size)
mat2 = torch.rand(size,size)
```

```python
start_time = time.time()
res =torch.matmul(mat1,mat2)
cpu_time = time.time() - start_time
print(f'time taken by cpu {cpu_time:.4f}')
```

```python
gpu_mat1 = mat1.to(device)
gpu_mat2 = mat2.to(device)

start_time = time.time()
res =torch.matmul(gpu_mat1,gpu_mat2)
torch.cuda.synchronize()
gpu_time = time.time() - start_time
print(f'time taken by gpu {gpu_time:.4f}')
```

```python
print('\nSpeedup (cpu/gpu):',cpu_time/gpu_time)
```
